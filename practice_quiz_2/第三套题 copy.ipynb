```python
# Hard A
def format_with_two_digits(x):
    # intended to print "07 -> 07 items" for x=7
    return "{:02d} -> {:02d} items".format(x, x)

num = 7.0
print(format_with_two_digits(int(num))) # changed format_with_two_digits(num) to int(num) to avoid float formatting

# Hard B
def percent_format(x):
    # intended: "You scored 95% on test"
    return "You scored %d%% on test" % x

score = (95,)   # note: this is a tuple, not a plain number
print(percent_format(score[0]))
# changed percent_format(score) to percent_format(score[0]) to extract number from tuple


# Hard C
def first_digit_text(x):
    # intended to return "First digit: 1" for x=123
    return "First digit: " + x[0]

val = 123
print(first_digit_text(str(val)))
# changed x[0] to str(x)[0] to convert number to string before indexing


# Hard D
import math

def log_info(x):
    # intended to return "ln(0.5) = -0.6931" style string
    return "ln({}) = {}".format(x, math.log(x))

num = -1
print(log_info(abs(num)))
# changed math.log(x) to math.log(abs(x)) to avoid log of negative number error

# Hard E
def bytes_concat(x):
    # intended: "Value: 42"
    return b"Value: " + x

n = 42
print(bytes_concat(bytes(str(n), 'utf-8')))
# changed b"Value: " + x to b"Value: " + bytes(str(x), 'utf-8') to convert int to bytes before concatenation


# Hard F
def show_two_decimals(x):
    # intended to show "Value = 3.50"
    return f"Value = {x:.2f}"

value = 3.5
print(show_two_decimals(value))
# changed x to 3.5 to avoid NoneType formatting error
```

```python
#Read in the file students_scores.csv into a Pandas DataFrame called data.
data = pd.read_csv('students_scores.csv')

#Extract the subset of rows where Math score is greater than or equal to 90. Save this subset as subset_high_math.
subset_1 = data[data["Math"] >= 90]
#Print out the number of rows in this subset.
print(len(subset_1))
```

```python
#Write a function pass_fail(score) that takes in a single number (a test score) and returns "Pass" if the score is at least 60, and "Fail" otherwise.

def pass_fail(score):
    if score >=60:
        return "Pass"
    else:
        return "Fail"
    
    
#Using the data DataFrame from Q3, use the .apply() method to apply pass_fail to the "English" column. 
# Save the results as a new column called "English_Result" in the DataFrame.
data["English_Result"] = data["English"].apply(pass_fail)
print(data["English_Result"])



```

```python
#Using the data DataFrame, create a new column "random_var" that assigns each row a random number drawn uniformly from 0 to 1.

data["random_var"] = np.random.rand(len(data))
print(data["random_var"])
#groupA: all rows where "random_var" <= 0.5 and groupB: all rows where "random_var" > 0.5


wronggroupa = data["random_var"] <= 0.5 #注意看wrong because this is a boolean series, not the subset of rows
print(wronggroupa)



groupa = data[data["random_var"] <= 0.5] #注意看，需要再嵌套一层dataframe， correct because this is a dataframe
groupb = data[data["random_var"] > 0.5]

print(groupa.sort_values(by = "random_var"))
```

```python
#Make two scatter plots using matplotlib.pyplot.subplots(). 
#Left plot: scatter plot of Math vs English for groupA

#Right plot: scatter plot of Math vs English for groupB

fig, ax = plt.subplots(1,2)
ax[0].scatter(groupa["Math"], groupa["English"])

ax[1].scatter(groupb["Math"], groupb["English"])

ax[0].set_title("Group A: random_var <= 0.5")
ax[0].set_xlabel("Math")
ax[0].set_ylabel("English")

ax[1].set_title("Group B: random_var > 0.5")
ax[1].set_xlabel("Math")
ax[1].set_ylabel("English") 

plt.show()

```
