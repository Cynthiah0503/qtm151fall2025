```python
data = pd.read_csv("/Users/cynthia/Documents/GitHub/qtm151fall2025/practice_quiz_2/students_scores.csv")
#取出第 200–250 行（包括 250 行）。
subset = data.iloc[200:251, ]
#按照 Biology 成绩从高到低排序，再按 Math 成绩从低到高作为次排序键。

#分开满足条件
biology = subset.sort_values(by="Biology")
print(biology.head(5))
math = biology.sort_values(by="Math", ascending=True)
print(math.head(5))

#同时满足俩条件应该这样，我上面写的不对
subset3 = subset.sort_values(by=["Biology", "Math"], ascending=[False, True])
print(subset3.head(5))
```

```python
#创建 subset1：筛选出 Age 在 18 到 20 岁（含边界）的行。
subset1 = data[ (data["Age"] >= 18)  & (data["Age"] <= 20)]


#创建 subset2：筛选出 Math 分数低于 60 或 English 分数高于 90 的行。

subset2 = data[ (data["Math"] < 60)  | (data["English"] > 90)]

print(subset1.sort_values(by="Age"))
print(subset2)
```

```python
#给每一行随机分配一个标签 "Red", "Green", "Blue", "Yellow"。

group = np.random.choice( ["Red", "Green", "Blue", "Yellow"], size=len(data) )
data["groupcolors"] = group

#画出散点图：横轴是 Math，纵轴是 English。

blue = data[ data["groupcolors"] == "Blue"]

plt.scatter( blue["Math"], blue["English"])
plt.xlabel("Math Scores")
plt.ylabel("English Scores")
plt.title("Math vs English Scores for Blue Group")
plt.show()


```
