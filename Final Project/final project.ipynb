```python
#rename the passing column without number to passing_16
df_merged = df_merged.rename(columns={"passing": "passing_16"})

#get the passing skill for year 2016 to 2022
passing_cols = [f"passing_{y}" for y in years]
print(passing_cols)

#create a new dataframe with only sofifa_id and passing skills
df_passing = df_merged[["sofifa_id"] + passing_cols]

df_passing = df_passing.dropna()

df_passing["mean_passing"] = df_passing[passing_cols].mean(axis=1)
#axis=1 means calculate the mean for each row


df_passing_sorted = df_passing.sort_values(by="mean_passing", ascending=False, inplace=False)

display(df_passing_sorted)


```

```python
#for plotting, convert from wide to long format
df_long = df_passing_sorted.melt(
    id_vars=["sofifa_id"], 
    var_name="year",     #year of the match
    value_name="score"   #passing skill score
)

#convert year from passing_16 to 2016

#remove the mean_passing column from the long dataframe
df_long = df_long[df_long["year"] != "mean_passing"]

df_long["year"] = df_long["year"].str.extract(r"(\d+)").astype(int) + 2000

display(df_long)

```

```python
#plot passing skills over years for all players
plt.figure(figsize=(10,6))
for pid in df_long["sofifa_id"].unique():
    df_player = df_long[df_long["sofifa_id"] == pid]
    plt.plot(df_player["year"], df_player["score"])

plt.title("Female Player Technical Trend (Passing 2016 to 2022)")
plt.xlabel("Year")
plt.ylabel("Passing Score")

#add a legend for the top 5 players with highest mean passing score
top5 = df_passing_sorted["sofifa_id"].head(5).values
for pid in top5:
    df_player = df_long[df_long["sofifa_id"] == pid]
    plt.plot(df_player["year"], df_player["score"], label=f"Player {pid}", linewidth=2.5)
plt.legend()


plt.show()
```

# progress made on passing skills from 2016 to 2022

```python
start = "passing_16"
end = "passing_22"
df_merged["overall_change"] = df_merged[end] - df_merged[start]

#fill na with 0
df_merged["overall_change"] = df_merged["overall_change"].fillna(0)

df_progress = df_merged[       ["sofifa_id", start, end, "overall_change"]      ].sort_values(by="overall_change", ascending=False)

display(df_progress)
```

```python
#make a graph showing overall change for all players

plt.figure(figsize=(10,6))
plt.bar(df_progress["sofifa_id"].astype(str), df_progress["overall_change"])
#.astype(str) will convert sofifa_id to string for better x axis display


plt.xticks(rotation=90, fontsize=8)
plt.title("Overall Passing Skill Change from 2016 to 2022 for Female Players")
plt.xlabel("Sofifa ID")
plt.ylabel("Overall Change in Passing Skill")
plt.show()
```
