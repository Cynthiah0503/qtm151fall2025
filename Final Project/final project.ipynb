# get all the passing score columns

```python
#rename the passing column without number to passing_16
df_merged = df_merged.rename(columns={"passing": "passing_16"})

#get the passing skill for year 2016 to 2022
passing_cols = [f"passing_{y}" for y in years]
print(passing_cols)

#create a new dataframe with only sofifa_id and passing skills
df_passing = df_merged[["sofifa_id"] + passing_cols]

df_passing = df_passing.dropna()

df_passing["mean_passing"] = df_passing[passing_cols].mean(axis=1)
#axis=1 means calculate the mean for each row


df_passing_sorted = df_passing.sort_values(by="mean_passing", ascending=False, inplace=False)

display(df_passing_sorted)


```

# make the dataframe into long form for graphing

```python
#for plotting, convert from wide to long format
df_long = df_passing_sorted.melt(
    id_vars=["sofifa_id"], 
    var_name="year",     #year of the match
    value_name="score"   #passing skill score
)

#convert year from passing_16 to 2016

#remove the mean_passing column from the long dataframe
df_long = df_long[df_long["year"] != "mean_passing"]

df_long["year"] = df_long["year"].str.extract(r"(\d+)").astype(int) + 2000

display(df_long)

```

## clean the dataset to get all the nationalities

```python
#rename the passing column without number to passing_16
df_merged = df_merged.rename(columns={"nationality_id": "nationality_id_16"})
df_merged = df_merged.rename(columns={"nationality_name": "nationality_name_16"})

#get the passing skill for year 2016 to 2022
nationality_id_cols = [f"nationality_id_{y}" for y in years]
print(nationality_id_cols)

nationality_name_cols = [f"nationality_name_{y}" for y in years]
print(nationality_name_cols)


#create a new dataframe with only sofifa_id and passing skills
df_nationality = df_merged[["sofifa_id"] + nationality_name_cols + nationality_id_cols]
df_nationality = df_nationality.dropna()


df_nationality
```

# find unique nationality counts

```python
#find all the unique nationalities in df_nationality
unique_nationalities = set()
for col in nationality_name_cols:
    unique_nationalities.update(df_nationality[col].unique())
print(unique_nationalities)


unique_nationalities = ['Mexico', 'France', 'England', 'Spain', 'Germany', 'United States', 'Australia', 'Sweden', 'Canada']


nationality_summary = {}

for nation in unique_nationalities:
    #find all players from a nation in any year
    mask = False
    for col in nationality_name_cols:     # ['nationality_name_16', ..., 'nationality_name_22']
        mask = mask | (df_nationality[col] == nation)     #the meaning of | is "or" operation for boolean masks

    #filter unique players from this nation
    players = df_nationality[mask]["sofifa_id"].unique()

    nationality_summary[nation] = {
        "count": len(players)
    }



#make into dataframe for better display
nationality_summary_df = pd.DataFrame.from_dict(nationality_summary, orient='index').reset_index()
nationality_summary_df.columns = ['Nationality', 'Count']
display(nationality_summary_df)



```

# merged passing score dataset with nationalities of players by sofifa id

```python
display(df_passing_sorted)
#merge df_passing_sorted with df_nationality on sofifa_id
df_nationality_passing = df_passing_sorted.merge(
    df_nationality,
    on='sofifa_id',
    how='left'
)


display(df_nationality_passing)

#melt df_nationality_passing to long format
df_nationality_long = df_nationality_passing.melt(
    id_vars=["sofifa_id", "mean_passing"],
    value_vars=nationality_name_cols,
    var_name="year",
    value_name="nationality"
)

display(df_nationality_long)


#find unique nationalities, and add a column for it
unique_nationalities = df_nationality_long["nationality"].unique()
```

# found germany have the best players with the highest average passing scores

```python
df_country_avg = (
    df_nationality_long[["sofifa_id", "mean_passing", "nationality"]]
    .drop_duplicates(subset=["sofifa_id"])           # only count each player once
    .groupby("nationality")                          # grouping
    .agg(avg_passing=("mean_passing", "mean"),       # aggregating
         count_players=("sofifa_id", "count"))       # additionally: count players per country
    .reset_index()
    .sort_values("avg_passing", ascending=False)     # sorting
)

display(df_country_avg)
```

# heights

```python
#rename the passing column without number to passing_16
df_merged = df_merged.rename(columns={"height_cm": "height_cm_16"})

#get the passing skill for year 2016 to 2022
height_cm_cols = [f"height_cm_{y}" for y in years]
print(height_cm_cols)


#create a new dataframe with only sofifa_id and passing skills
df_height_cm = df_merged[["sofifa_id"] + height_cm_cols]
df_height_cm = df_height_cm.dropna()


# find a mean height_cm for all players, average the numbers from 16 to 22
df_height_cm_mean["mean_height_cm"] = df_height_cm[height_cm_cols].mean(axis=1)

#define a function to first round to nearest integer, and convert into meters
def cm_to_meters(cm):
    cm_rounded = round(cm)
    meters = cm_rounded / 100
    return meters

df_height_cm_mean["mean_height_meters"] = df_height_cm_mean["mean_height_cm"].apply(cm_to_meters)

display(df_height_cm_mean)


```

# categorize the heights into tall and short

#### https://pmc.ncbi.nlm.nih.gov/articles/PMC9795175/
# data show that mean female FIFA player's height is 1.65

#### Furthermore, the average height of females ages 20–29 in the world is 162.56 cm (5′4″)

```python
# define a function to classify players by height
def classify_height(meters):
    if meters < 1.625:
        return "short"
    else:
        return "tall"

# use apply to create a new categorical column
df_height_cm_mean["height_category"] = df_height_cm_mean["mean_height_meters"].apply(classify_height)

display(df_height_cm_mean)
```

# relationship between height and skill level(passing score)

## there is little or no relationship between height and passing skills

```python
#merge df_passing_sorted with df_height_cm_mean on sofifa_id
df_passing_height = df_passing_sorted.merge(
    df_height_cm_mean,
    on='sofifa_id',
    how='left'
)

display(df_passing_height)


# group short vs high and compute mean passing score
df_height_passing_summary = (
    df_passing_height
    .groupby("height_category")["mean_passing"]
    .mean()
    .reset_index()
    .rename(columns={"mean_passing": "avg_passing_score"})
)

display(df_height_passing_summary)
```
